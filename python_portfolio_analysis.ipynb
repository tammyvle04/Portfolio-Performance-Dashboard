```python
import yfinance as yf

tickers = ["AAPL", "JPM", "JNJ", "XOM", "PG", "NVDA", "SPY"]

data = yf.download(tickers, start = "2021-01-01")

data.head()
```

```python
close_prices = data["Close"]

close_prices.head()
```

```python
close_prices.to_csv('stock_prices.csv')
```

```python
# DAILY RETURNS

import pandas as pd

close_prices = pd.read_csv('stock_prices.csv', index_col=0, parse_dates = True)


daily_returns = close_prices.pct_change().dropna()
daily_returns.head()
```

```python
# CUMULATIVE RETURNS (how much $1 invested in the beginning would be worth over time)

cumulative_returns = (1 + daily_returns).cumprod() - 1
cumulative_returns.tail()
```

```python
# VOLATILITY

rolling_volatility = daily_returns.rolling(window=30).std()
rolling_volatility.tail()
```

```python
# CORRELATION MATRIX (SPY excluded due to calculation amongst holdings, not benchmark)

holdings = ["AAPL", "JPM", "JNJ", "XOM", "PG", "NVDA"]
correlation_matrix = daily_returns[holdings].corr()
correlation_matrix
```

```python
# PORTFOLIO VS. BENCHMARK ($10k in holdings vs $10k in benchmark)

portfolio_cumulative = cumulative_returns[holdings].mean(axis=1)
benchmark_cumulative = cumulative_returns['SPY']

portfolio_value = 10000 * (1 + portfolio_cumulative)
benchmark_value = 10000 * (1 + benchmark_cumulative)


comparison = pd.DataFrame({
    'Portfolio ($10k Invested)' : portfolio_value,
    'S&P 500 ($10k Invested)' : benchmark_value
})
comparison.tail()
```

```python
daily_returns.to_csv('daily_returns.csv')
cumulative_returns.to_csv('cumulative_returns.csv')
correlation_matrix.to_csv('correlation_matrix.csv')
comparison.to_csv('portfolio_vs_benchmark.csv')
rolling_volatility.to_csv('rolling_volatility.csv')
                          
```
